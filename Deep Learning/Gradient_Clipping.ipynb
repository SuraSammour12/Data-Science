✄ Gradient clipping

Gradient clipping is a technique used to prevent exploding gradients in neural networks by capping the gradients during backpropagation to a predefined threshold.

If the magnitude of a gradient exceeds the threshold, it is scaled down to ensure it stays within the limit.

Benefits:
Stabilizes training, especially in recurrent neural networks (RNNs) and deep architectures.
Helps maintain the numerical stability of the optimization process.


---
 two techniques to control gradient magnitudes: clipnorm and clipvalue

 **1. clipnorm (Gradient Norm Clipping)**

This method ensures that the entire gradient vector does not exceed a certain norm (magnitude).

If the norm of the gradient is larger than the specified clipnorm, it is scaled down proportionally to match the clipnorm value.

Formula:

gradient=gradient×(clipnorm/(max(clipnorm,||gradient||)))

Effect: Keeps gradients under control and prevents sudden, extreme updates to model parameters

Example:

If clipnorm=1, all gradients are normalized so their total magnitude does not exceed 1

--------------------------------------------------------------------------
**2. clipvalue (Gradient Value Clipping)**

This method limits individual gradient components to a fixed range.

Any gradient value above clipvalue is set to +clipvalue, and any value below -clipvalue is set to -clipvalue.

Formula:

gradient[i]=max(−clipvalue,min(gradient[i],clipvalue))

Effect: Prevents exploding gradients by capping extreme values in the gradient vector

Example:

If clipvalue=1, any gradient element greater than 1 is set to 1, and any gradient element less than -1 is set to -1


---
clipnorm: Limits the entire gradient vector’s norm and scales down the whole gradient proportionally.

clipvalue: Restricts individual gradient values to prevent extreme gradient components.


                   


```python
# clipvalue
optimizer = keras.optimizers.SGD(clipvalue=1.0)
```

```python
# clipnorm
optimizer = keras.optimizer.SGD(clipnorm=1.0)
```
