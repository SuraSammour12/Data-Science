**Faster optimizers**

Faster optimizers are advanced optimization algorithms designed to speed up the convergence of neural networks during training.

They improve efficiency by adapting learning rates or using momentum to navigate the loss landscape effectively.

Popular in deep learning tasks, such as image recognition or natural language processing, where faster convergence is crucial for large datasets.


---



👉 **Momentum optimization**

Momentum: Adds a fraction of the previous update to the current update to maintain velocity in the right direction, the algorithm introduces a new hyperparameter β, simply called the momentum, which must be set between 0 (high friction) and 1 (no friction). A typical momentum value is 0.9.


Linking Problem Solving 📖 with Momentum in Optimization 🚀

🔹 Slow Solution Approach: Searching for page 400 by flipping pages one by one is slow and inefficient (similar to linear search).

🔹 Smarter Approach: Open the book in the middle, then take large steps before refining the search, making it faster by using previous knowledge.

🔹 **Momentum in Optimization:** Works similarly by leveraging past direction to speed up convergence and avoid obstacles.

Key Equation:
Wnew = Wold-LR⋅∇G+(last update × fraction)

                                        =>  (new argument)
                       


---



```python
# Momentum optimization
# The code sets up an SGD optimizer with a learning rate of 0.001 for slow, stable updates and momentum of 0.9 to speed up training and reduce oscillations.(by incorporating past updates)
optimizer = keras.optimizers.SGD(learning_rate=0.001, momentum=0.9)
# learning_rate=0.001 → Controls the step size for weight updates.
# momentum=0.9 → Helps accelerate learning by incorporating previous updates.
```



---


👉 **Nesterov Accelerated Gradient**

```python
# Nesterov Accelerated Gradient
# This optimizer uses SGD with Nesterov momentum, which predicts future positions before updating, leading to faster and more stable learning
optimizer = keras.optimizers.SGD(learning_rate=0.001, momentum=0.9, nesterov=True)
# learning_rate=0.001 → Controls the step size for weight updates
# momentum=0.9 → Helps accelerate learning by incorporating previous updates
# nesterov=True → Instead of blindly applying momentum, Nesterov looks ahead before updating, leading to more precise and stable optimization
```

👉 **RMSProp**

It's like scaling the learning rate

Wnew=Wold−LR⋅∇G

It adjusts the learning rate for each parameter by taking the average of the magnitude of the previous gradients

When to Use It?

* With deep neural networks, especially in image and speech processing.

* Ideal for LSTMs and time-series tasks due to its stability in handling varying gradients



```python
# RMSProp
# This code initializes the RMSProp optimizer, an advanced gradient-based optimization algorithm
optimizer = keras.optimizers.RMSprop(learning_rate=0.001, rho=0.9)
# learning_rate=0.001 → Controls how much weights are updated
# rho=0.9 → Determines how much past gradients influence the current update, helping to smooth updates
```



---
👉 **Adam Optimization**

Combines the features of Momentum and RMSProp:

Uses momentum to update the direction.

Uses RMSProp to adapt the learning rate based on the moving average of past gradients.

Wnew=Wold−LR⋅∇G+(last update×fraction)

                (scaling + new argument )

Benefits:

* Faster convergence compared to traditional gradient descent

* Works well in most deep learning applications without requiring fine-tuning of the learning rate parameters

When to Use It?

* Ideal for most deep learning applications.

* Works well in neural networks, computer vision, and NLP tasks.
                    
                               🕸️     👁    🗣️🤖📜  


```python
# Adam Optimization
# This code initializes the Adam optimizer, one of the most powerful and widely used optimization algorithms in deep learning
optimizer = keras.optimizers.Adam(learning_rate=0.001, beta_1=0.9, beta_2=0.999)
# learning_rate=0.001 → Controls the weight update step size
# beta_1=0.9 → Controls momentum, storing past gradients to accelerate learning
# beta_2=0.999 → Controls RMSProp, adjusting learning rates for individual parameters
# Adam combines the benefits of both Momentum and RMSProp, making it both fast and adaptive
```



---
📌 **In brief:**

**Momentum** helps maintain the correct direction

**RMSprop** adapts the learning rate based on past gradients

**Adam** combines them for the best performance and fastest convergence

👉 **Adamax Optimization**

Difference between Adam and Adamax :

Adamax uses the maximum of past gradients instead of the average, which makes it more stable when gradients are large or unstable.


Why Adamax?

✅ Higher stability: Performs better when gradients are fluctuating or large

✅ Retains Adam's advantages: Works well in most deep learning applications, with better stability in some cases

When to Use It?

Used in the same situations as Adam, but may offer better results when gradients are unstable




```python
# Adamax Optimization
# This code initializes the Adamax optimizer, a variant of the Adam algorithm,
# with adjustments in the way gradients are handled, offering more stability in certain cases
optimizer = keras.optimizers.Adamax(learning_rate=0.001, beta_1=0.9, beta_2=0.999)
# learning_rate=0.001 → Defines the step size for weight updates.
# beta_1=0.9 → Controls momentum, storing past gradients to speed up learning
# beta_2=0.999 → Controls RMSProp, adjusting the learning rate based on past squared gradients
```

👉 **Nadam Optimization**

Nadam uses Nesterov momentum, which makes updates more accurate before actually applying them, improving training stability and speeding up convergence.

Why Nadam?

✅ Proactive updates: With Nesterov momentum, updates are improved before making final decisions, speeding up training and reducing oscillations

✅ Powerful hybrid: Performs well across a variety of deep learning problems due to its combination of Adam and NAG

✅ Excellent performance: Provides better stability and faster performance compared to some other optimizers

When to Use It?

Commonly used in deep neural networks and deep learning in general

Performs well in image processing and natural language processing tasks

                      📷🛠️                💬🛠️                   



```python
# Nadam Optimization
# This code initializes the Nadam optimizer, which combines the benefits of Nesterov Accelerated Gradient (NAG) and Adam
optimizer = keras.optimizers.Nadam(learning_rate=0.001, beta_1=0.9, beta_2=0.999)
# learning_rate=0.001 → Defines the step size for weight updates
# beta_1=0.9 → Controls momentum, storing past gradients to accelerate learning
# beta_2=0.999 → Controls RMSProp, adjusting the learning rate based on past squared gradients
```
