🔹 **Sigmoid Activation Function**

The Sigmoid Activation Function is a nonlinear activation function used in neural networks that transforms any input value into a range between 0 and 1

🔹 **Mathematical Equation:**

𝜎(𝑥)=1/1+e^-x




```python
# Import libraries

import numpy as np
# numpy is used for numerical operations


np.random.seed(42)
# Set random seed for reproducibility

# To plot pretty figures 🎀
%matplotlib inline
import matplotlib as mpl
import matplotlib.pyplot as plt
mpl.rc('axes',labelsize=14)
mpl.rc('xtick',labelsize=12)
mpl.rc('ytick',labelsize=12)
```

```python
# Define the Sigmoid function
# logit(z) returns values between 0 and 1 following the sigmoid formula
def logit(z):
  return 1/(1+np.exp(-z)) # This is sigmoid activation function formula , z is parameter and it's the input of this function
```

```python
# Create input values for plotting , it's the input for logit
z = np.linspace(-5,5,200) #  # Generate values from -5 to 5

plt.plot([-5,5],[0,0],'k-')
plt.plot([-5,5],[1,1],'k--')
plt.plot([0,0],[-0.2,1.2],'k-')
plt.plot([-5,5],[-3/4,7/4],'g--')

plt.plot(z,logit(z),'b-',linewidth=2) # # Plot Sigmoid function
props = dict(facecolor='black',shrink=0.1)
# Annotate important regions (Saturation & Linear)
# Labels different parts of the sigmoid function.
plt.annotate('Saturating',xytext=(3.5,0.7),xy=(5,1), arrowprops=props,fontsize=14,ha='center')
plt.annotate('Saturating',xytext=(-3.5,0.3),xy=(-5,0), arrowprops=props,fontsize=14,ha='center')
plt.annotate('Linear',xytext=(2,0.2),xy=(0,0.5), arrowprops=props,fontsize=14,ha='center')
# Enhances the graph with a title and grid.
# Displays the final plot.
plt.grid(True)
plt.title('Sigmoid Activation Function',fontsize=14)
plt.axis([-5,5,-0.2,1.2])
plt.show()
```

here, I can see at the final output figure the **"Active Region"**

🔹 What is the Active Region? 👩‍🏫

The Active Region of an activation function is the range where the output significantly changes in response to input variations. In this region, the function is most sensitive to changes in input values.

🔹 Active Region in the Sigmoid Function

In the figure, we can see that the Active Region of the Sigmoid function is approximately between -4 and 4

Inside this range (-4 to 4): The output gradually transitions from 0 to 1

Outside this range (less than -4 or greater than 4): The output saturates, meaning it becomes almost 0 or 1, and further input changes have little to no effect

For example:

✅ If the input is very large (like 50 or 500), the output will be very close to 1

✅ If the input is very small (like -50 or -500), the output will be very close to 0


🔴 Important Note:

Saturated activation functions like Sigmoid and Tanh suffer from the **saturation problem**, leading to poor learning at large values. Therefore, **ReLU** is often preferred to avoid this issue
