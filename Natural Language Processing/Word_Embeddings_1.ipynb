 **Goal of the code:**

The purpose is to train a Word2Vec model using Gensim on a small sample of sentences, to:

Build a vocabulary from the text

Learn word embeddings (vector representations)

Find semantically similar words using cosine similarity


---




```python
# Import libraries after restart
import gensim
from gensim.models import Word2Vec
import logging

# Enable logging to track training progress
logging.basicConfig(format="%(asctime)s : %(levelname)s : %(message)s", level=logging.INFO)

# Sample corpus
sentences = [
    ["the", "cat", "sat", "on", "the", "mat"],
    ["the", "dog", "barked", "at", "the", "cat"],
    ["the", "cat", "chased", "the", "mouse"],
    ["dogs", "and", "cats", "can", "be", "friends"],
]

# Train the Word2Vec model
model = Word2Vec(sentences, vector_size=50, window=3, min_count=1, workers=4, sg=1)
"""
Key parameters:
vector_size=50: Each word will be represented as a 50-dimensional vector
window=3: Context window size (how many words to the left/right to consider)
min_count=1: Keep words that appear at least once
sg=1: Use Skip-Gram architecture (better for rare words)
workers=4: Use 4 CPU cores for faster training

"""

# Show vocabulary
print("Vocabulary:", list(model.wv.index_to_key))

# Word embedding for "cat"
print(f"\nEmbedding for 'cat':\n", model.wv["cat"])
print("Length of embedding:", len(model.wv["cat"]))

# Find most similar words to "cat"
similar_words = model.wv.most_similar("cat", topn=5)
print(f"\nMost similar words to 'cat':\n", similar_words)

# Save and reload model
model.save("word2vec.model")
loaded_model = Word2Vec.load("word2vec.model")
print("\n✅ Model loaded successfully.")
```
