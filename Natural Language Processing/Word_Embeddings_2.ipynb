
This code uses the gensim library to build a Word2Vec model in two different modes:

CBOW (Continuous Bag of Words): Predicts a target word based on its surrounding context.

Skip-gram: Does the opposite — it tries to predict context words given a target word.

The model is trained on a text file called "alice.txt"

```python
# Installing and loading spaCy:
!pip install -U spacy
!python -m spacy download en_core_web_sm

import spacy
from google.colab import files

nlp = spacy.load("en_core_web_sm")


# Uploading the file:
uploaded = files.upload()

# Reading the text and analyzing sentences and words:
filename = list(uploaded.keys())[0]
with open(filename, 'r', encoding='utf-8') as sample:
    text = sample.read()


doc = nlp(text)

# Converting sentences to lists of words:
data = []
for sent in doc.sents:
    temp = [token.text.lower() for token in sent if not token.is_space]
    data.append(temp)


print("First sentence :", data[0])
```

Word2Vec models

CBOW (Continuous Bag of Words)
Skip-Gram
⭢ to analyze input text and extract similarities between words based on context

Tools Used:

spaCy: A natural language processing (NLP) library used to analyze text and extract information like sentences, tokens, and other linguistic features. It’s used here to process the uploaded text and identify sentences and words

Gensim: A library used to build and train word embeddings such as Word2Vec, which transforms words into numerical vectors that can be used to measure word similarity

```python
!pip install -U spacy gensim
!python -m spacy download en_core_web_sm

import spacy
import warnings
from gensim.models import Word2Vec
from google.colab import files

warnings.filterwarnings(action='ignore')


nlp = spacy.load("en_core_web_sm")

# Uploading the file:
uploaded = files.upload()


filename = list(uploaded.keys())[0]
with open(filename, 'r', encoding='utf-8') as file:
    text = file.read()


text = text.replace("\n", " ")

doc = nlp(text)
data = []
for sent in doc.sents:
    temp = [token.text.lower() for token in sent if not token.is_space]
    data.append(temp)

# A CBOW model is trained on the processed sentences. In this model, the context of surrounding words is used to predict the target word.
# Training the CBOW model
model1 = Word2Vec(sentences=data, vector_size=100, window=5, min_count=1)

# After training, cosine similarity between words is computed using the similarity function
print("Cosine similarity between 'alice' and 'wonderland' - CBOW:",
      model1.wv.similarity('alice', 'wonderland'))

print("Cosine similarity between 'alice' and 'machines' - CBOW:",
      model1.wv.similarity('alice', 'machines'))


# The Skip-Gram model is trained similarly, but here the target word is used to predict its surrounding context words
# Training the Skip-Gram model
model2 = Word2Vec(sentences=data, vector_size=100, window=5, min_count=1, sg=1)

# After training, cosine similarity between words is computed using the similarity function
print("Cosine similarity between 'alice' and 'wonderland' - Skip Gram:",
      model2.wv.similarity('alice', 'wonderland'))

print("Cosine similarity between 'alice' and 'machines' - Skip Gram:",
      model2.wv.similarity('alice', 'machines'))
```

**CBOW Model Results:**

* The very high similarity (0.99) between 'alice' and 'wonderland' shows that these two words often appear in similar contexts in the text — which makes sense 🤔, since Alice is in Wonderland! 👧🏼 📎 °⋆🎡.ೃ✨࿔*･🎪

* 'alice' and 'machines' also have fairly high similarity (0.93). This might be due to both words appearing in somewhat related sentence structures, or the text might contain machines or machine-like concepts (maybe "thinking machine", "dream machine", etc.).👧🏼 📎 🧐🖥 , 👧🏼 📎 😴✨🌌💫🖥

**Skip-Gram Model Results:**


*   Similar trend, but scores are slightly lower than in CBOW
*   This is expected: CBOW tends to be more stable and smooth, while Skip-Gram is better for rare words and more precise but noisier.

📘 **Sentence:**

"Alice saw a white rabbit running through the forest."

Let’s focus on the word "rabbit"



---
🧩**1. How does CBOW work?**
(Uses context to predict the target word)

Idea:
CBOW looks at the surrounding words around "rabbit" — for example: "a", "white", "running", "through" — and tries to predict the missing word in the middle, which is "rabbit".

It's like the model is asking:
If I have: "a white ... running through", what word probably fits in the middle?

✅ It learns that "rabbit" often appears in this type of context.



---

🧩 **2. How does Skip-Gram work?**
(Uses the target word to predict the surrounding words)

Idea:
Skip-Gram takes the word "rabbit" and asks:

Based on this word, what are the likely words that appear around it?

✅ It tries to predict words like:

"white" ✅

"forest" ✅

"computer" ❌ (unlikely)

So the model learns that "rabbit" is commonly found near words like "white", "run", and "forest".


---
**📌 Summary:**

**Model	| How it thinks | 	Strengths | 	Weaknesses**

CBOW	| "Give me the context, I'll predict the word" | 	Fast, stable |	Not great with rare words

Skip-Gram |	"Give me the word, I'll predict the context" |	Better with rare or specific words	 | Can learn weird associations (noisy)
