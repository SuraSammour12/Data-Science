**Sentence Segmentation** ✄ 💬

Splits text into individual sentences

**Purpose:**

Facilitates tasks like summarization and sentiment analysis
Ensures proper tokenization and focus on sentence-level meaning





---
**nltk (Natural Language Toolkit)** library for segmenting text into sentences

**Punkt** is a pre-trained model used by nltk to segment text into sentences and words

**sent_tokenize()** is a function from the nltk library that uses linguistic analysis techniques (such as rules and trained models) to accurately segment text


```python
import nltk
nltk.download('punkt_tab')
from nltk.tokenize import sent_tokenize

# Example text
text = "Hi I'm Sura. I'm learning data science now"

# Divide the text into sentences
sentences = sent_tokenize(text)

print(sentences)
```
