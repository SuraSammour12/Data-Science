**Special Tokens in Tokenizers**

Natural language tokenizers use special tokens to handle certain tasks such as padding, classification, and separating segments. Below are their meanings:

[PAD]: Used to make all input sequences the same length by adding padding at the end.

[UNK]: Represents unknown words not found in the vocabulary.

[CLS]: Placed at the beginning of every input for classification tasks.

[SEP]: Separates multiple sentences or segments in one input.

[MASK]: Used in training to hide a word so the model learns to predict it.

[unusedX]: Reserved tokens not currently used but available for custom use.

```python
from transformers import BertTokenizer
```

```python
# The code loads the BERT tokenizer and analyzes it
tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")

def analyze_tokenizer(tokenizer):
    # Print the total vocabulary size in the form (the number of words it knows)
    vocab_size = tokenizer.vocab_size
    print(f"Vocabulary size: {vocab_size}")
    # Show the first 10 symbols in the vocabulary, just for a look 👀
    sample_vocab = list(tokenizer.vocab.keys())[:10]
    print(f"Sample of the vocabulary: {sample_vocab}")
    # Print special characters and their IDs within vocabulary
    print("Special tokens:")
    print(f" - [PAD] token: {tokenizer.pad_token}, ID: {tokenizer.pad_token_id}")
    print(f" - [CLS] token: {tokenizer.cls_token}, ID: {tokenizer.cls_token_id}")
    print(f" - [SEP] token: {tokenizer.sep_token}, ID: {tokenizer.sep_token_id}")
    print(f" - [MASK] token: {tokenizer.mask_token}, ID: {tokenizer.mask_token_id}")
    # A text example that we pass to the encoder to convert it to encoded tokens (input_ids)
    sample_text = "Pizza is my  weakness. No regrets"
    encoded = tokenizer(sample_text)
    # Print Original Text
    # Print Tokenization
    # Print Input_ids
    print(f"\nSample text: {sample_text}")
    print(f"Tokenized text: {tokenizer.tokenize(sample_text)}")
    print(f"Encoded IDs: {encoded['input_ids']}")
    # Convert the digital identifiers back to text to see how the model understands the text
    decoded_text = tokenizer.decode(encoded['input_ids'])
    print(f"Decoded text: {decoded_text}")

# Run the analysis
analyze_tokenizer(tokenizer)
```

**Result after encryption (Encoded IDs):**

Encoded IDs: [101, 10733, 2003, 2026, 11251, 1012, 2053, 23161, 102]

This is the text after converting it to numbers (IDs), Where:

101 = [CLS]

102 = [SEP]

The others are the IDs for the words

 **Result after decryption (Decoded text):**

 Decoded text: [CLS] pizza is my weakness. no regrets [SEP]

 => The model was able to successfully convert numbers to words again, automatically adding the symbols [CLS] and [SEP]

here we can see the **Case Normalization**

the BERT tokenizer uses a model called "bert-base-uncased", which performs case normalization. This means that all capital letters (uppercase) in the input text are converted to lowercase
