The following code aims to parse and tokenize Arabic text using the **tiktoken library**, which is used to understand how text is divided into "tokens" in **GPT models** like GPT-3 and GPT-3.5/4

A **token** is a small unit of text (which could be a word, a part of a word, or even a symbol like "?"), and **language models like ChatGPT** deal with tokens, not raw text

The TikToken library developed by OpenAI

This library uses the same slicing techniques as the real GPT


---



Import the library , write Arabic text


```python
import tiktoken
arabic_text = "مرحبًا, كيف حالك اليوم؟ هذا نص باللغة العربية."
```

Select the type of slicing based on the model:

```python
gpt3_tokenizer = tiktoken.get_encoding('r50k_base')
gpt3_5_gpt4_tokenizer = tiktoken.get_encoding('cl100k_base')
# r50k_base: This is the type of slicing used by GPT-3 models
# cl100k_base: This is the newer type used in GPT-3.5 and GPT-4
```

Slicing text into tokens

```python
gpt3_tokens = gpt3_tokenizer.encode(arabic_text)
gpt3_5_gpt4_tokens = gpt3_5_gpt4_tokenizer.encode(arabic_text)
# encode() converts text into a list of numbers (tokens)
# Each number represents a unit of text (a word, part of a word, or a symbol)
```

Print results:

```python
print("Text:", arabic_text)
print("Tokens (GPT-3):", len(gpt3_tokens), gpt3_tokens)
print("Tokens (GPT-3.5/GPT-4):", len(gpt3_5_gpt4_tokens), gpt3_5_gpt4_tokens)
# Prints the number of tokens used for each model
# It also prints a list of tokens (numbers representing text)
```

📊 **Model Comparison:**



---
**Model | Number of Tokens |  Notes**

GPT-3 (r50k_base) |  48 tokens | is too many, because the hashing is old and not optimized for Arabic

GPT-3.5/GPT-4 (cl100k_base) | 36 tokens | fewer because the slicing is better and it supports multiple languages ​​




---
**Why does this matter?**

When dealing with OpenAI APIs, the cost of consumption is calculated based on the number of tokens, as follows:

Long text input = more tokens = higher cost

Long text output = more tokens = higher cost

Therefore, fewer tokens for the same meaning = better performance and lower cost
