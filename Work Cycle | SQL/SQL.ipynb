###  **Data Science Cycle**

---

###  1. **Business**
The process starts with a business request for an idea or project, such as predicting demand or analyzing customer behavior.

---

###  2. **Data Engineer**
**Design Schema:** Defines tables, columns, and relationships.  
Example: Name = character type, max length 15.  

**Pipelines:** Build data pipelines to extract from multiple sources, clean them, and unify their structure.  

**ETL (Extract, Transform, Load):**  
- **Extract:** Get data from sources.  
- **Transform:** Clean and standardize.  
- **Load:** Store it into the Data Warehouse.  

**Data Warehouse:** Central storage for clean data.  

💡 **SQL:** Used heavily by Data Engineers; others use it to query data.

---

###  3. **Data Analyst**
Creates reports and dashboards (e.g., Power BI).  
Analyzes past data to give insights.  

💡 Often uses Excel/CSV; may use SQL when needed.


---

###  4. **Data Scientist (ML/DL)**
**Preprocessing**, **Training**, **Tuning**, and **Evaluation** using:  
RMSE, R², F1 Score ..

Saves and publishes models to predict future outcomes.  

💡 Sometimes uses SQL for data access.

---

###  5. **DevOps / MLOps**
Deploy models for use in apps/APIs.  
Ensures smooth, stable ML model operations.


```python
"""
Data Science Cycle 📊🤖🔁

Business --> Request for idea/project

DE --> Design Schema | Pipelines | ETL | Data Warehouse --> SQL

DA --> Reports and Dashboards (Power BI) --> SQL (if needed)

DS (ML/DL) --> Preprocessing | Training | Tuning | Evaluation (RMSE, R2, F1) | Save & Publish --> SQL (sometimes)

DevOps / MLOps --> Deploy the model

"""
```

### 🧪 **Basic SQL Commands with SQLite (just simple example)**
These examples simulate working with a small SQL database inside Google Colab using `sqlite3`.

```python
# Step 1: Import SQLite and create in-memory database
import sqlite3

# Create a connection
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()
```

```python
# Step 2: Create a sample table
cursor.execute('''
CREATE TABLE students (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    age INTEGER,
    grade TEXT
);
''')
```

🔔 These are simple examples of applying CRUD (Create, Read, Update, Delete) operations at the record level.

```python
# Step 3: Insert some sample data
cursor.executemany('''
INSERT INTO students (name, age, grade) VALUES (?, ?, ?)
''', [
    ('Alice', 22, 'A'),
    ('Bob', 23, 'B'),
    ('Charlie', 21, 'C')
])
conn.commit()
```

```python
# Step 4: SELECT - Show all data
cursor.execute('SELECT * FROM students')
for row in cursor.fetchall():
    print(row)
```

```python
# Step 5: UPDATE - Change Bob's grade to A
cursor.execute("UPDATE students SET grade = 'A' WHERE name = 'Bob'")
conn.commit()
```

```python
# Step 6: DELETE - Remove Charlie from the table
cursor.execute("DELETE FROM students WHERE name = 'Charlie'")
conn.commit()
```

```python
# Final SELECT - Check updated table
cursor.execute('SELECT * FROM students')
for row in cursor.fetchall():
    print(row)
```
